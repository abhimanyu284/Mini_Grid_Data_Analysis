```python
import os
import pandas as pd

LONDON = "../data/raw/london"
BLOCKS = os.path.join(LONDON, "hhblock_dataset", "hhblock_dataset")

hh = pd.read_csv(os.path.join(LONDON, "informations_households.csv"))
hol = pd.read_csv(os.path.join(LONDON, "uk_bank_holidays.csv"))
wx_d = pd.read_csv(os.path.join(LONDON, "weather_daily_darksky.csv"))
wx_h = pd.read_csv(os.path.join(LONDON, "weather_hourly_darksky.csv"))

for name, df in {"households": hh, "holidays": hol, "weather_daily": wx_d, "weather_hourly": wx_h}.items():
    print("=" * 60)
    print(name, df.shape)
    print(df.dtypes)
    print("missing:\n", df.isna().sum())
    print("duplicates:", df.duplicated().sum())
    display(df.head(3))
```

```python
print(hh.columns.tolist())
for col in hh.columns:
    if hh[col].nunique() < 30:
        print("\n", col)
        print(hh[col].value_counts(dropna=False))
```

```python
block = pd.read_csv(os.path.join(BLOCKS, "block_0.csv"))
print(block.shape)
print(block.columns.tolist()[:6], "...", block.columns.tolist()[-2:])
print("households in block:", block.iloc[:, 0].nunique())
print("missing values:", block.isna().sum().sum())
print("duplicates:", block.duplicated().sum())
display(block.head(3))
```

```python
date_col = [c for c in block.columns if "day" in c.lower()][0]
block[date_col] = pd.to_datetime(block[date_col])
print("date range:", block[date_col].min(), "->", block[date_col].max())

files = [f for f in os.listdir(BLOCKS) if f.startswith("block_")]
total_mb = sum(os.path.getsize(os.path.join(BLOCKS, f)) for f in files) / 1e6
print(len(files), "blocks,", round(total_mb), "MB total")
```
